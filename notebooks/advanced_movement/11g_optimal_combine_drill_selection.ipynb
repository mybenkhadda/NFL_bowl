# 11G — Optimal Combine drill selection

**Question (S11G).** Which subset of Combine tests provides the greatest reliable, incremental information for NFL movement evaluation?

**Objectives, reported separately** (no invented scouting utility): held-out prediction on a COMMON cohort · repeatability · redundancy with official tests · physical coverage of NFL demands · participation (missing-data resilience). Complexity control: each drill family enters a model as 3 principal components fitted on the training athletes of the fold. The order in which drills are added is chosen inside each outer fold (honest saturation curve).

**Missingness.** Non-forty participation is nonrandom; a participation model shows how. A complete-case gain is never generalised to all prospects.

**Costs** are DECLARED HYPOTHETICAL minutes, not measurements.

**Prerequisites.** 11.0, 11A, 11C.

```python
RUN_MODE = 'full'
SEED = 20261011
FORCE_REBUILD = False
SAVE_TABLES = True
SAVE_FIGURES = True
NUM_WORKERS = 0
TARGETS = ('nfl.band_4_6', 'nfl.brake_decel', 'nfl.reaccel_mean', 'nfl.accel_low_mean')   # declared prediction objectives
MAX_FAMILIES = 6
```

```python
from nflbdb.data.config import Phase01Config, configure_runtime

cfg = Phase01Config.load()
configure_runtime(cfg)

import numpy as np
import polars as pl
from IPython.display import Markdown, display

from nflbdb.research.advanced_movement import config as amc
from nflbdb.research.advanced_movement.artifacts import DiscoveryStore
from nflbdb.research.advanced_movement.m11 import assemble as asm
from nflbdb.research.advanced_movement.m11 import config as m11c
from nflbdb.research.advanced_movement.m11 import drill_selection as ds
from nflbdb.research.advanced_movement.m11 import evaluation as ev
from nflbdb.research.advanced_movement.m11 import figures as fg
from nflbdb.research.advanced_movement.m11 import reporting as rp
from nflbdb.research.advanced_movement.m11 import sources as src
from nflbdb.research.advanced_movement.m11 import tournament as tn

controls = amc.NotebookControls(run_mode=RUN_MODE, force_rebuild=FORCE_REBUILD, save_figures=SAVE_FIGURES, save_tables=SAVE_TABLES,
                                seed=SEED, num_workers=NUM_WORKERS).validate()
store = DiscoveryStore(cfg, controls)
amc.seed_everything(controls.seed)
C = m11c.M11Config().for_mode(controls)
PROV = m11c.provenance('11G', controls, C)
W = store.write_frame
data = asm.linked_table(store)
registry = store.upstream.require_named('athlete_registry', src.NB['00'])
idx = asm.foundation(store, 'm11_attempt_index')
fm = asm.foundation(store, 'm11_feasibility_matrix')
rep = asm.foundation(store, 'm11_repeatability')
families = [f for f in src.FAMILY_ORDER if f in set(idx['family'].to_list()) and f != 'other'][:MAX_FAMILIES]
FAM = ds.family_columns(data.columns, families)
label = 'SMOKE — wiring check, numbers are not results' if controls.smoke else 'FULL'
display(Markdown(f'**Run mode: `{controls.run_mode}`** ({label}) · families: {families}'))
display(pl.DataFrame([{'family': f, 'n_features': len(v)} for f, v in FAM.items()]))
```

## 1 · Participation and missingness (G4)

```python
att_ath = idx.select('nfl_id', 'family').unique()
part = ds.participation_model(registry, att_ath, families)
W(part['rates'], 'm11_participation_rates')
W(part['coefficients'], 'm11_participation_coefs')
with pl.Config(tbl_rows=-1, tbl_width_chars=200):
    display(part['coefficients'])
rates_pos = part['rates'].filter(pl.col('by') == 'position_group')
store.show(fg.fig_heatmap(rates_pos, 'family', 'level', 'rate', 'Drill participation by position (missingness)',
                          'Share of each position’s Combine athletes with ≥ 1 usable tracked attempt of the drill. Participation is nonrandom: complete-case results describe participants only.', fmt='{:.2f}'),
           'AM11_G_01_participation', {'rates': rates_pos}, publication=True, source='m11_participation_rates')
```

## 2 · Prediction on common cohorts: marginal value, random-family control, nested selection curve (G3)

For each declared target, the families with enough linked participants are compared on the athletes having ALL of them (common cohort). If the common cohort is too small, the families with the fewest participants are dropped (reported) until it reaches the support line.

```python
marg, curve, orders, cohorts = [], [], [], []
for tgt in TARGETS:
    if tgt not in data.columns:
        cohorts.append({'target': tgt, 'families': '', 'n_common': 0, 'status': 'target_not_available'})
        continue
    fams = {f: c for f, c in FAM.items() if c}
    while fams:
        cc = ds.common_cohort(data, tgt, fams)
        if cc.height >= C.min_athletes_cell or len(fams) == 1:
            break
        drop = min(fams, key=lambda f: data.filter(pl.any_horizontal([pl.col(k).is_not_null() for k in fams[f]]) & pl.col(tgt).is_not_null()).height)
        fams.pop(drop)
    cc = ds.common_cohort(data, tgt, fams) if fams else data.clear()
    cohorts.append({'target': tgt, 'families': ', '.join(fams), 'n_common': cc.height, 'status': 'computed' if cc.height >= C.min_athletes_limited else 'too_few'})
    if cc.height < C.min_athletes_limited:
        continue
    out = ds.selection_experiment(cc, tgt, fams, cc['fold'].to_numpy(), C, C.seed)
    marg.append(out['marginal']); curve.append(out['curve']); orders.append(out['orders'])
COH = pl.DataFrame(cohorts)
MARG = pl.concat(marg, how='diagonal_relaxed') if marg else pl.DataFrame()
CURVE = pl.concat(curve, how='diagonal_relaxed') if curve else pl.DataFrame()
ORD = pl.concat(orders, how='diagonal_relaxed') if orders else pl.DataFrame()
for k, v in (('m11_drill_cohorts', COH), ('m11_drill_marginal', MARG), ('m11_drill_curve', CURVE), ('m11_drill_orders', ORD)):
    W(v, k)
with pl.Config(tbl_rows=-1, tbl_width_chars=240):
    display(COH)
    display(MARG.select('target', 'family', 'n_athletes', 'delta_rmse', 'delta_lo', 'delta_hi', 'verdict') if MARG.height else MARG)
    display(CURVE.select('target', 'n_added_families', 'n_athletes', 'delta_rmse', 'delta_lo', 'delta_hi', 'verdict') if CURVE.height else CURVE)
if MARG.height:
    mf = MARG.with_columns((pl.col('target') + ' · ' + pl.col('family').str.replace('_random', 'RANDOM (control)')).alias('label'))
    store.show(fg.fig_forest(mf, 'label', 'delta_rmse', 'delta_lo', 'delta_hi', 'Incremental information of each drill beyond official tests',
                             'Held-out Δ RMSE of official tests + one drill family (3 training-fold principal components) vs official tests only, on the common cohort of each target; RANDOM = 3 noise columns (control).',
                             'Δ RMSE (target units)', color_by='target'), 'AM11_G_02_incremental_waterfall', {'marginal': mf}, publication=True, source='m11_drill_marginal')
if CURVE.height:
    cf = CURVE.with_columns((pl.col('target') + ' · +' + pl.col('n_added_families').cast(pl.String) + ' drills').alias('label'))
    store.show(fg.fig_forest(cf, 'label', 'delta_rmse', 'delta_lo', 'delta_hi', 'Information saturation (nested forward selection)',
                             'Δ RMSE vs official tests when the first k drills chosen INSIDE each outer fold are added (honest held-out curve).', 'Δ RMSE', color_by='target'),
               'AM11_G_03_saturation', {'curve': cf}, publication=True, source='m11_drill_curve')
if ORD.height:
    display(ORD.filter(pl.col('rank') == 1).group_by('target', 'family').len().sort('target', 'len', descending=[False, True]))
```

## 3 · Objectives table, redundancy, coverage frontier and position-specific orderings (G5)

```python
red_rows = []
for f, cols in FAM.items():
    for k in cols[:12]:
        red_rows.append({'metric': k, **ev.stopwatch_redundancy(data, k)})
RED = pl.DataFrame(red_rows, infer_schema_length=None) if red_rows else pl.DataFrame()
obj = ds.objectives_table(families, MARG, rep, RED, fm, part['rates'])
front = ds.coverage_frontier(fm, families)
W(RED, 'm11_drill_redundancy')
W(obj, 'm11_drill_objectives')
W(front, 'm11_drill_frontier')
with pl.Config(tbl_rows=-1, tbl_width_chars=260):
    display(obj)
store.show(fg.fig_frontier(front, 'Drill selection frontier (coverage vs hypothetical cost)',
                           'Each point = a drill subset (forty always included). y = SUPPORTED NFL demand × position cells with ≥ medium declared overlap; x = DECLARED HYPOTHETICAL minutes per athlete (not measured).'),
           'AM11_G_04_frontier', {'frontier': front}, publication=True, source='m11_drill_frontier')
cov = fm.filter((pl.col('feasibility') == m11c.SUPPORTED) & pl.col('overlap_confidence').is_in(['high', 'medium'])).group_by('family', 'demand_id').len()
store.show(fg.fig_heatmap(cov, 'family', 'demand_id', 'len', 'NFL physical demand coverage by drill', 'Number of positions for which the drill × demand cell is SUPPORTED with ≥ medium overlap.'),
           'AM11_G_05_coverage_matrix', {'coverage': cov}, publication=False, source='m11_feasibility_matrix')
if RED.height:
    rr = RED.drop_nulls('r2_official').with_columns(pl.col('metric').str.split('.').list.first().alias('family'), pl.col('metric').str.split('.').list.last().alias('feature'))
    store.show(fg.fig_heatmap(rr, 'feature', 'family', 'r2_official', 'Redundancy with official tests', 'Within-position R² of each drill feature from the official forty and 10-yd split (1 = fully explained by the stopwatch).', fmt='{:.2f}'),
               'AM11_G_06_redundancy', {'redundancy': rr}, publication=False, source='m11_drill_redundancy')
pos_rows = []
for (pos,), g in sorted(fm.partition_by(['position_group'], as_dict=True).items()):
    pr = part['rates'].filter((pl.col('by') == 'position_group') & (pl.col('level') == pos))
    o = ds.objectives_table(families, pl.DataFrame(), rep, RED, g, pr)
    pos_rows.append(o.with_columns(pl.lit(pos).alias('position_group')))
POS = ds.position_recommendations(pl.concat(pos_rows, how='diagonal_relaxed')) if pos_rows else pl.DataFrame()
W(POS, 'm11_drill_position_orderings')
display(POS)
```

```python
evid = []
for r in (MARG.filter(pl.col('family') != '_random').iter_rows(named=True) if MARG.height else []):
    rnd = MARG.filter((pl.col('target') == r['target']) & (pl.col('family') == '_random'))
    evid.append(tn.evidence_row('family.drill_information', '11G', position_group=r['family'], n_athletes=r['n_athletes'], nfl_comparison=r['target'], delta_rmse=r['delta_rmse'],
                                delta_lo=r['delta_lo'], delta_hi=r['delta_hi'], beats_controls=bool(rnd.height and r['delta_rmse'] is not None and rnd['delta_rmse'][0] is not None and r['delta_rmse'] > rnd['delta_rmse'][0]),
                                execution_mode=controls.run_mode, config_hash=C.hash(), note='position_group column holds the drill family here'))
W(tn.evidence_frame(evid), 'm11_metric_evidence_11G')
verd = MARG.filter(pl.col('family') != '_random')['verdict'].to_list() if MARG.height else []
status = rp.comparison_status(verd, computed=bool(MARG.height))
facts = {'status': status, 'run_mode': controls.run_mode,
         'sample': '; '.join(f"{r['target']}: {r['n_common']} athletes with {r['families'] or 'no family'}" for r in COH.iter_rows(named=True)),
         'evidence': [f"{r['target']} + {r['family']}: Δ RMSE {rp.ci(r['delta_rmse'], r['delta_lo'], r['delta_hi'], 4)} ({r['verdict']})" for r in (MARG.iter_rows(named=True) if MARG.height else [])]
                     + [f"frontier: {r['subset']} covers {r['covered_cells']} cells at {r['hypothetical_cost_min']:.0f} hypothetical min" for r in front.filter(pl.col('pareto')).iter_rows(named=True)][:6],
         'alternatives': ['Common cohorts are small and select athletes who did every drill.', 'A drill can be valuable for a demand that has no reliable NFL target yet (coverage without prediction).'],
         'relevance': 'Separates drills that add held-out information, drills that only add physical coverage, and drills that duplicate the stopwatch — per objective, without a combined score.',
         'limitations': ['Hypothetical costs.', 'Participation is nonrandom; position-specific orderings rest on few athletes.'],
         'next': ['Collect the most informative non-forty drills for more athletes before deciding.', 'Re-run on a later class with drills frozen.']}
md_ = rp.conclusion_markdown('11G', facts)
display(Markdown(md_))
rp.write_conclusion(store, '11G', md_)
rp.run_manifest(store, '11G', PROV, ['m11_drill_marginal', 'm11_drill_curve', 'm11_drill_objectives', 'm11_drill_frontier', 'm11_participation_coefs'], status)
crow = pl.DataFrame([{'item': f"{r['target']} + {r['family']}", 'evidence': f"Δ RMSE {rp.ci(r['delta_rmse'], r['delta_lo'], r['delta_hi'], 4)}, n={r['n_athletes']}", 'status': rp.comparison_status([r['verdict']])}
                     for r in (MARG.filter(pl.col('family') != '_random').iter_rows(named=True) if MARG.height else [])])
store.show(fg.fig_conclusion(crow, '11G — conclusion', 'Marginal held-out value of each drill family per target (common cohorts).'), 'AM11_G_07_conclusion', {}, publication=True, source='m11_drill_marginal')
print('Send back: participation coefficients, cohorts, marginal and curve tables, objectives, frontier, conclusion.')
print('Next: 11h_position_specific_football_actions.ipynb')
```
